## 分かったことの読み方

計算上の徒歩到達範囲は、歩行速度・出発点・OSMの道路種別と接続に依存します。接続線は実際の入口ではなく、駅の改札や建物内部の経路も確認していません。待ち時間・坂・階段の負荷・混雑は含みません。

「駅から徒歩10分で行けるカフェを、この仮定で数える」のように目的を限定すると説明しやすくなります。POI登録数は営業中店舗数ではなく、売上・需要・出店可否・住みやすさの判断には使いません。吉祥寺1駅から一般化しません。

今回の結果：1km四方3,591件、円3,481件、徒歩5分2,141件、徒歩10分3,794件。円の3,481件はすべて今回の徒歩10分条件に入りました。一方、正方形だけに入る12件と徒歩10分だけに入る215件がありました。範囲の違いを一方向の包含関係とは決めつけられません。
