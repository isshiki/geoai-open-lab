# 地図のメッシュ入門
標準地域メッシュ・H3・S2・独自グリッドを比較します。どれも同じ「500m」ではありません。

[方法と公式出典](../docs/mesh-systems.md) / [検証](../docs/mesh-systems-validation.md)

準備：`uv sync --locked --all-groups`。以降のセルはオフラインで実行できます。

```python
from geoai_open_lab.mesh_systems import run, OUT
from geoai_open_lab.mesh_guide_figures import render
from IPython.display import SVG, display

comparison = run()
comparison.query("city == '東京・吉祥寺'")[["scheme", "level", "cell_id", "area_km2", "child_count", "neighbor_count"]]
```

## 同じ地点が入る区画を並べる
黒点は同じ代表地点。全パネルは同縮尺で北が上です。面積はWGS84楕円体上の境界面積。投影座標上の幅500mとは区別します。

```python
render()
display(SVG(filename=str(OUT / "boundaries.svg")))
```

## 同じレベルでも面積は変わる
札幌・那覇は比較用に固定した地点で、駅の位置ではありません。独自グリッドはEPSG:6677の適用域を考慮し東京だけに計算しています。

```python
comparison.pivot(index="scheme", columns="city", values="area_km2").round(6)
```

## 何のために区切るか
公的統計とコードで結合するなら標準地域メッシュ。XYZとQuadkeyは同一タイルの別表現。H3の親子は論理的階層で、子の境界が親に厳密に収まるとは限りません。OGC DGGSは枠組みで、ここでは適合性を認証していません。

```python
display(SVG(filename=str(OUT / "choice.svg")))
```
